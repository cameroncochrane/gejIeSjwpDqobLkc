<h1>MonReader: Clip Modelling (CM): Data Processing and Model Establishment

The clip model (CM) is essentially extending the single-frame classifier (SFM) from spatial feature learning to spatio-temporal feature learning.

The biggest conceptual change is that the input is no longer one image: (H, W, C),

but a short sequence, one clip: (T, H, W, C),

where T is the number of frames in the clip. 

- The SFM answers 'What does this frame look like?'

- A CM needs to answer 'What changes across these frames, and does that sequence look like a page flip?'

For the CM, temporal information could be extremely useful because a page flip is inherently a motion event rather than purely an appearance.

TensorFlow has temporal functionalities, one of them being 'TimeDistributed', which can be integrated into a CNN similar to the ones being used during the SFM part of the project.
